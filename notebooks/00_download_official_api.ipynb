# 00 — Download Kroger-family supermarkets from the official API

Use OAuth2 client credentials to collect Kroger and Kroger-owned supermarket banners from the official Chains and Locations APIs. Create `.env` from `.env.example` first. Secrets are never displayed or stored in this notebook.

```python
from pathlib import Path
import sys

def find_project_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / "scripts" / "build_all.py").exists():
            return candidate
    raise FileNotFoundError("Run this notebook from inside the repository.")

ROOT = find_project_root()
sys.path.insert(0, str(ROOT / "scripts"))
ROOT
```

## Credential setup

Register an application at https://developer.kroger.com/. Copy `.env.example` to `.env`, then enter `KROGER_CLIENT_ID` and `KROGER_CLIENT_SECRET`. The `.env` file is excluded from Git.

```python
import subprocess, sys

print('[1/2] Starting official Kroger-family supermarket collection...')
result = subprocess.run([sys.executable, str(ROOT / 'scripts' / 'download_official_api.py')], cwd=ROOT)
if result.returncode:
    raise RuntimeError('Official API collection failed. Check .env credentials and the progress messages above.')
print('[2/2] Official family collection complete.')
```

```python
import pandas as pd
official = pd.read_csv(ROOT / 'data' / 'processed' / 'kroger_family_official_locations.csv', dtype={'zip_code': 'string'})
print(f'{len(official):,} unique official API supermarket locations across {official.brand.nunique()} banners')
official.head()
```
