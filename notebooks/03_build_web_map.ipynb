# 03 — Build the interactive web map

Generate `docs/index.html`, which embeds the location points and can be published from GitHub Pages. No machine-specific file paths are stored in the output.

```python
from pathlib import Path
import sys

def find_project_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / "scripts" / "build_all.py").exists():
            return candidate
    raise FileNotFoundError("Run this notebook from inside the repository.")

ROOT = find_project_root()
sys.path.insert(0, str(ROOT / "scripts"))
ROOT
```

```python
import pandas as pd
from build_all import build_map

print('[1/4] Loading official family location and state-summary files...')
locations = pd.read_csv(ROOT / 'data' / 'processed' / 'kroger_family_official_locations.csv', dtype={'zip_code': 'string'}).fillna('')
state_summary = pd.read_csv(ROOT / 'data' / 'processed' / 'state_summary.csv')
print(f'[2/4] Loaded {len(locations):,} locations across {locations.brand.nunique()} banners and {len(state_summary)} states.')
print('[3/4] Building the interactive web map; this can take a moment...')
build_map(locations, state_summary)
print('[4/4] Complete: created docs/index.html')
```
